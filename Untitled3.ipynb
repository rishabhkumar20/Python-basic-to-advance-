```python
#here's is an examle of set:
my_set = { 1, 2, 3}
```

```python
# adding elements 
my_set.add(4)
print(my_set)
```

```python
#removing element 
my_set.remove(1)
print(my_set)
```

```python
#union of sets:
other_set ={4, 5, 6}
union_set = my_set.union(other_set)
print(union_set)
```

```python
#intersection of sets:
intersection_set = my_set.intersection(other_set)
print(intersection_set)


                    
```

```python
# difference of set: Elements that are present in the first set but NOT present in the second set.
difference_set = my_set.difference(other_set)
print(difference_set)
```

```python
#one key features of set is that they automatically remove dupplicate value 
duplicate_set = {1, 1, 2, 2, 3, 3}
print(duplicate_set)
```
